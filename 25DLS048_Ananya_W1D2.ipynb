```python
import numpy as np
import matplotlib.pyplot as plt

#Create matrix A(3X2) and B(2X3) with random integers
A=np.random.randint(0,10,(3,2))
B=np.random.randint(0,10,(2,3))
print("Matrix A:\n", A)
print("Matrix B:\n", B)

#Multiply A and B
C=np.dot(A,B)
print("Matrix A x B:\n", C)

# Define matrix D
D = np.array([[4, 2], [1, 3]])

# Find eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(D)
print("Eigenvalues:\n", eigenvalues)
print("Eigenvectors:\n", eigenvectors)

# Plot eigenvectors using quiver
origin = np.zeros((2, 2))  # origin point for vectors
plt.figure(figsize=(6,6))
plt.quiver(*origin, eigenvectors[0], eigenvectors[1], angles='xy', scale_units='xy', scale=1, color=['r','b'])
plt.xlim(-1.5, 1.5)
plt.ylim(-1.5, 1.5)
plt.grid()
plt.title("Eigenvectors of matrix D")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
import sympy as sp

# Define symbolic variable and function
x = sp.symbols('x')
f = 3*x**3 - 5*x**2 + 2*x - 1

# Compute derivative
df = sp.diff(f, x)
print("Derivative f'(x):", df)

# Evaluate derivative at x=2
df_at_2 = df.subs(x, 2)
print("Derivative at x=2:", df_at_2)

# Define function f(x, y)
x, y = sp.symbols('x y')
f_xy = x**2 + y**2

# Compute gradient
grad_f = [sp.diff(f_xy, var) for var in (x, y)]
print("Gradient:", grad_f)

# Gradient descent step from (3, 4) with learning rate 0.1
current_point = {x: 3, y: 4}
learning_rate = 0.1

grad_at_point = [g.subs(current_point) for g in grad_f]
print("Gradient at (3,4):", grad_at_point)

# Update x and y
new_x = current_point[x] - learning_rate * grad_at_point[0]
new_y = current_point[y] - learning_rate * grad_at_point[1]
print(f"New point after gradient descent step: ({new_x}, {new_y})")
```

```python
# Dice simulation
dice_rolls = np.random.randint(1, 7, 1000)

# Plot histogram
plt.hist(dice_rolls, bins=np.arange(1, 8)-0.5, edgecolor='black')
plt.title("Histogram of 1000 Dice Rolls")
plt.xlabel("Dice Face")
plt.ylabel("Frequency")
plt.xticks(range(1, 7))
plt.show()

# Empirical mean and variance
mean_dice = np.mean(dice_rolls)
var_dice = np.var(dice_rolls)
print(f"Empirical mean of dice rolls: {mean_dice}")
print(f"Empirical variance of dice rolls: {var_dice}")

# Biased coin toss simulation (P(heads) = 0.7)
coin_outcomes = ['Heads', 'Tails']
probabilities = [0.7, 0.3]
coin_tosses = np.random.choice(coin_outcomes, 500, p=probabilities)

# Count heads and tails
counts = {side: np.sum(coin_tosses == side) for side in coin_outcomes}
print("Counts of coin tosses:", counts)

# Plot bar chart
plt.bar(counts.keys(), counts.values(), color=['blue', 'orange'])
plt.title("Biased Coin Toss Counts (500 tosses)")
plt.ylabel("Count")
plt.show()
```
